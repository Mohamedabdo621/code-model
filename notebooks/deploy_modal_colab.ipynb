# نشر النموذج على Modal — رابط دائم (Deploy to Modal)

النوتبوك دي بتنشر نموذجك على Modal مرة واحدة، وبتطلع لك **رابط دائم**
مش بيخلص ولا بيتغير — بتحطه في التطبيق مرة واحدة وبتنساه.

## قبل ما تشغّل (مرة واحدة بس، دقيقتين)
1. اعمل حساب على https://modal.com — سجّل دخول بـ GitHub.
2. افتح https://modal.com/settings/tokens ودوس **Create new token**.
3. هيظهر لك قيمتين: **Token ID** بيبدأ بـ `ak-` و **Token Secret** بيبدأ
   بـ `as-`. انسخهم وخليهم جاهزين — هتلصقهم في خلية الدخول اللي تحت
   (السر بيظهر مرة واحدة بس، ومتشاركهوش مع أي حد ولا في الشات).

وبعدين شغّل الخلايا دي بالترتيب (Run all بيكفي).


```python
# 1) Install Modal CLI
!pip install -q modal
import modal
print("✅ Modal installed:", modal.__version__)
```

```python
# 2) Log in with your API token
# بيحاول يقرأ الأسرار MODAL_TOKEN_ID / MODAL_TOKEN_SECRET من Colab Secrets الأول،
# ولو مش موجودة هيطلب منك تلصق القيمتين هنا (الإدخال مخفي).
import os

token_id = token_secret = None
try:
    from google.colab import userdata
    token_id = userdata.get("MODAL_TOKEN_ID")
    token_secret = userdata.get("MODAL_TOKEN_SECRET")
except Exception:
    pass

if token_id and token_secret:
    print("✅ Logged in via Colab Secrets")
else:
    import getpass
    print("الصق الـ Token ID (بيبدأ بـ ak-):")
    token_id = getpass.getpass("Token ID: ").strip()
    print("الصق الـ Token Secret (بيبدأ بـ as-):")
    token_secret = getpass.getpass("Token Secret: ").strip()

os.environ["MODAL_TOKEN_ID"] = token_id
os.environ["MODAL_TOKEN_SECRET"] = token_secret
print("✅ Token ready — الخلية الجاية هتنشر")
```

```python
# 3) Download the deploy file from GitHub and deploy (النشر)
# النشر الأول بياخد كام دقيقة عشان بيئة التشغيل بتتبني — سيبه يخلص.
import re
import subprocess
import sys
import urllib.request

urllib.request.urlretrieve(
    "https://raw.githubusercontent.com/Mohamedabdo621/code-model/main/modal_app.py",
    "modal_app.py",
)
print("⬇️ modal_app.py downloaded — deploying now...")

proc = subprocess.Popen(
    [sys.executable, "-m", "modal", "deploy", "modal_app.py"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
)
output_lines = []
for line in proc.stdout:
    print(line, end="")
    output_lines.append(line)
proc.wait()

urls = re.findall(r"https://[^\s\]]+\.modal\.run", "".join(output_lines))
if proc.returncode == 0 and urls:
    print("\n" + "=" * 60)
    print("🎉 اتنشر! رابط الـ API الدائم بتاعك هو:")
    print(urls[0])
    print("=" * 60)
    print("ابعته لـ Pip في الشات عشان يختبره، وبعدين الصقه في التطبيق")
    print("في تبويب «الرابط» مرة واحدة واحفظ.")
else:
    print("\n❌ النشر مكملش — انسخ آخر سطور من الخطأ وابعتها لـ Pip.")
```

## بعد النشر
